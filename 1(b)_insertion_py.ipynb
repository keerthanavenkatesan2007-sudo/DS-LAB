```python
class node:
  def __init__(self,dataval=None):
      self.dataval=dataval
      self.nextval=None
class slinkedlist:
  def __init__(self):
      self.headval=None
  def listprint(self):
      printval=self.headval
      while printval is not None:
           print (printval.dataval)
           printval=printval.nextval
  def atbeginning(self,newdata):
    newnode=node(newdata)
    newnode.nextval=self.headval
    self.headval=newnode

list_obj=slinkedlist()
list_obj.headval=node("mon")
e2=node("tue")
e3=node("wed")
list_obj.headval.nextval=e2
e2.nextval=e3
list_obj.atbeginning("sun")
list_obj.listprint()
```
