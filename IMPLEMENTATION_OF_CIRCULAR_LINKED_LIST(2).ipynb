CREATION

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CreateList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)
        if self.head is None:  # List is empty
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head  # Point to itself for a single node circular list
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head  # Connect new tail to head

    def display(self):
        if self.head is None:
            print("List is empty")
            return
        else:
            print("Nodes of the circular linked list:")
            current = self.head
            while True:
                print(current.data)
                current = current.next
                if current == self.head:  # Stop when we loop back to head
                    break

# Create an instance of CreateList and add elements
cl = CreateList()
cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)

# Display the circular linked list
cl.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.last = None

    def addToEmpty(self, data):
        if self.last is not None:
            # If the list is not empty, this function should not be called or should handle it.
            # For now, let's assume it's only called when empty.
            # Or, we can simply return the existing last node.
            return self.last

        temp = Node(data)
        self.last = temp
        self.last.next = self.last # Points to itself for a single node circular list
        return self.last

    def addBegin(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        return self.last

    def addEnd(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        self.last = temp
        return self.last

    def addAfter(self, data, item):
        if self.last is None:
            return None

        p = self.last.next # Start from the first node
        while True:
            if p.data == item:
                temp = Node(data)
                temp.next = p.next
                p.next = temp
                if p == self.last:
                    self.last = temp
                return self.last
            p = p.next
            if p == self.last.next:
                # Looped through all elements and didn't find the item
                print(item, "not present in the list")
                break
        return self.last

    def traverse(self):
        if self.last is None:
            print("List is empty")
            return

        temp = self.last.next # Start from the first node
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.last.next: # Stop when we loop back to the first node
                break
        print()

if __name__ == '__main__':

    llist = CircularLinkedList()

    last = llist.addToEmpty(6)
    last = llist.addBegin(4)
    last = llist.addBegin(2)
    last = llist.addEnd(8)
    last = llist.addEnd(12)
    last = llist.addAfter(10, 8)

    llist.traverse()
```

DELETION


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CreateList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)
        if self.head is None:  # List is empty
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head  # Points to itself for a single node circular list
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head  # Connect new tail to head

    def deleteEnd(self):
        if self.head is None:  # List is empty
            print("List is empty, nothing to delete.")
            return

        if self.head == self.tail:  # Only one node in the list
            self.head = None
            self.tail = None
            return

        current = self.head
        # Traverse until current.next is the tail
        while current.next != self.tail:
            current = current.next

        # Now 'current' is the node before the tail
        current.next = self.head  # Connect current to head, effectively removing tail
        self.tail = current      # Update tail to current

    def display(self):
        if self.head is None:
            print("List is empty")
            return
        else:
            print("Nodes of the circular linked list:")
            current = self.head
            while True:
                print(current.data)
                current = current.next
                if current == self.head:  # Stop when we loop back to head
                    break
            print() # Add a newline for better formatting

if __name__ == '__main__':
    cl = CreateList()
    cl.add(1)
    cl.add(2)
    cl.add(3)
    cl.add(4)

    print("Original List:")
    cl.display()

    print("Deleting from end:")
    while cl.head is not None:
        cl.deleteEnd()
        print("Updated List:")
        cl.display()
```
